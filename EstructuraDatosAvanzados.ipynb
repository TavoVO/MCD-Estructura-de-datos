```python
def contar(numero):
    if numero == 0:          # Caso base: detenerse
        return
    else:
        print(numero)
    contar(numero - 1) 
    
contar(3)      # Llamada recursiva (se llama a sí misma)
```

```python
def factorial(n):
    if n == 1:          # Caso base
        return 1
    else:
        print(f"Calculando factorial de {n}, llamando a factorial({n-1})")
        res = n * factorial(n - 1)
        print(res)
        return res  # Llamada recursiva
factorial(3) #3x2x1=6
```

```python
factorial(5) #5x4x3x2x1=120
```

```python
def fibonacci(n):
    if n == 0:
        return 0      # caso base 1
    elif n == 1:
        return 1      # caso base 2
    else:
        print(f"Calculando fibonacci de {n}, llamando a fibonacci({n-1}) y fibonacci({n-2})")
        return fibonacci(n-1) + fibonacci(n-2)  # llamada recursiva
    
print(fibonacci(6))  # 8
```

Determinar si dos arreglos de n posiciones son iguales

```python
# Arreglos
ar1 = [1, 2, 3, 4, 5]
ar2 = [1, 2, 3, 4, 5]

def son_iguales(ar1, ar2, i=0):
    if len(ar1) != len(ar2):
        return False  # Si las listas no tienen la misma longitud, no son iguales
    if i == len(ar1):
        return True  # Si se han comparado todos los elementos y son iguales
    if ar1[i] != ar2[i]:
        return False  # Si algún elemento es diferente, las listas no son iguales
    return son_iguales(ar1, ar2, i + 1)  # Llamada recursiva para comparar el siguiente elemento

son_iguales(ar1, ar2)  # True si son iguales o False si son diferentes
```

Calcular el número de ocurrencias del elemento elem en un arreglo


```python
arr = [1, 2, 1, 4, 5]
def ocurrencias(arr, elem, i=0):
    if i == len(arr):
        return 0  # Caso base: si se recorrió todo el arreglo
    if arr[i] == elem:
        return 1 + ocurrencias(arr, elem, i + 1)  # Sumar 1 si el elemento coincide
    return ocurrencias(arr, elem, i + 1)  # Continuar con el siguiente elemento

ocurrencias(arr, 1) # Calcula las ocurrencias del número 1
```

Decidir si un elemento elem aparece en un arreglo.

```python
arr = [3, 5, 2, 8, 6]
def aparece(arr, elem, i=0):
    if i == len(arr):
        return False  # Caso base: si se recorrió todo el arreglo
    if arr[i] == elem:
        return True  # Si el elemento coincide
    return aparece(arr, elem, i + 1)  # Continuar con el siguiente elemento

aparece(arr, 5) # si el número 5 aparece en el arreglo
```

Eliminar todas las ocurrencias del elemento elem en una lista encadenada

```python
arr = [1, 2, 1, 4, 5]  
def ocurrencias(arr, elem, i=0):
    if i == len(arr):
        return arr  # Caso base: si se recorrió todo el arreglo
    if arr[i] == elem:
        arr[i] = 0  # se reemplazo con 0 si coincide
    return ocurrencias(arr, elem, i + 1)  # Continuar con el siguiente elemento

ocurrencias(arr, 1)  
```

Tarea *Calcular la potencia entera de un número real*

```python
def potencia(num, pot, i=0):
    if i == pot:
        return 1  # Caso base: cualquier número elevado a la potencia 0 es 1
    return num * potencia(num, pot, i + 1)  # Llamada recursiva

potencia(2, 3)
```

```python
#invertir una cadena
def invertir_cadena(cadena, cad2="", i=0):
    if i == len(cadena):  # Caso base: cuando se recorre toda la cadena
        return cad2
    else:
        cad2 += cadena[len(cadena) - 1 - i]  # Concatenar el carácter correspondiente
        return invertir_cadena(cadena, cad2, i + 1)
    
invertir_cadena("hola")  # aloh
```

```python
#calcular la suma de los elementos de un arreglo
arr = [1, 2, 3, 4, 5]
def suma_arreglo(arr, n, sum=0):
    #print("n:",n)
    if n==0:
        return sum
    else:
        #print(arr[n-1])
        sum = sum + arr[n-1]
        #print("sum:",sum)
    return suma_arreglo(arr, n-1, sum)

suma_arreglo(arr,len(arr))  
#print(len(arr))# 15
```

```python
class Nodo:
    def __init__(self, tarea):
        self.tarea = tarea # dato del nodo en este caso la tarea
        self.siguiente = None # referencia al siguiente nodo

class Lista_enlazada:
    def __init__(self):
        self.cabeza = None # referencia al primer nodo de la lista

    def agregar_tarea(self, tarea): # agrega una tarea al final de la lista
        nuevo_nodo = Nodo(tarea)
        if self.cabeza is None: # si la lista está vacía
            self.cabeza = nuevo_nodo
        else:
            actual = self.cabeza
            while actual.siguiente: # recorrer hasta el final de la lista
                actual = actual.siguiente
            actual.siguiente = nuevo_nodo

    def agregar_tarea_inicio(self, tarea): # agrega una tarea al inicio de la lista
        nuevo_nodo = Nodo(tarea)
        nuevo_nodo.siguiente = self.cabeza
        self.cabeza = nuevo_nodo

    def contar_tareas(self): # cuenta el número de nodos en la lista
        contador = 0
        actual = self.cabeza
        while actual:   # recorrer la lista
            contador += 1
            actual = actual.siguiente 
        return contador

    def buscar_tarea(self, tarea): # busca una tarea en la lista
        actual = self.cabeza
        while actual: # recorrer la lista
            if actual.tarea == tarea:
                return True
            actual = actual.siguiente
        return False
    
    def mostrar_tareas(self): # muestra todas las tareas en la lista
        actual = self.cabeza
        while actual: # recorrer la lista
            print(actual.tarea)
            actual = actual.siguiente

    def eliminar_tarea(self, tarea): # elimina una tarea de la lista
        actual = self.cabeza
        previo = None
        while actual: # recorrer la lista
            if actual.tarea == tarea: # tarea encontrada
                if previo:
                    previo.siguiente = actual.siguiente
                else:
                    self.cabeza = actual.siguiente
                return True
            previo = actual
            actual = actual.siguiente
        return False
    
    def eliminar_ultima_tarea(self): # elimina la última tarea de la lista
        if self.cabeza is None:
            return False
        if self.cabeza.siguiente is None:
            self.cabeza = None
            return True
        actual = self.cabeza
        previo = None
        while actual.siguiente:
            previo = actual
            actual = actual.siguiente
        previo.siguiente = None
        return True
    
    def eliminar_primera_tarea(self): # elimina la primera tarea de la lista
        if self.cabeza is None: # si la lista está vacía
            return False
        self.cabeza = self.cabeza.siguiente # mover la cabeza al siguiente nodo
        return True
    
    
    def agregar_tarea_poscion(self, tarea, posicion): # agrega una tarea en una posición específica
        nuevo_nodo = Nodo(tarea)
        if posicion == 0: # si la posición es 0, agregar al inicio
            nuevo_nodo.siguiente = self.cabeza
            self.cabeza = nuevo_nodo
            return
        actual = self.cabeza # nodo actual
        previo = None
        indice = 0
        while actual and indice < posicion: # recorrer hasta la posición deseada
            previo = actual
            actual = actual.siguiente
            indice += 1
        previo.siguiente = nuevo_nodo # insertar el nuevo nodo
        nuevo_nodo.siguiente = actual # enlazar el nuevo nodo con el siguiente


# Instancia de la lista enlazada
lista = Lista_enlazada()
# Agregar tareas a la lista
lista.agregar_tarea("Tarea 1")
lista.agregar_tarea("Tarea 2")
lista.agregar_tarea("Tarea 4")
lista.agregar_tarea("Tarea 5")
lista.agregar_tarea_inicio("Tarea 0")

print("Número de nodos en la lista:", lista.contar_tareas())

print("¿Existe 'Tarea 1'?:", lista.buscar_tarea("Tarea 1"))

print("¿Existe 'Tarea 3'?:", lista.buscar_tarea("Tarea 3"))

print("Nodos en la lista:")
lista.mostrar_tareas()

lista.eliminar_tarea("Tarea 1")
print("Nodos en la lista después de eliminar 'Tarea 1':")
lista.mostrar_tareas()

lista.agregar_tarea_poscion("Tarea 3", 1)
print("Nodos en la lista después de agregar 'Tarea 3' en la posición 1:")
lista.mostrar_tareas()

lista.eliminar_ultima_tarea()
print("Nodos en la lista después de eliminar la última tarea:")
lista.mostrar_tareas()

lista.eliminar_primera_tarea()
print("Nodos en la lista después de eliminar la primera tarea:")
lista.mostrar_tareas()
```
