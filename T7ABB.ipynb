```python
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

class ABB:
    def __init__(self):
        self.root = None

    def list_to_binary_tree(self, arr, index=0):
        if index >= len(arr):
            return None
        
        root = Node(arr[index])
        root.left = self.list_to_binary_tree(arr, 2 * index + 1)
        root.right = self.list_to_binary_tree(arr, 2 * index + 2)
        return root
    
    def imprimir_arbol(self, node, level=0):
        if node is not None:
            self.imprimir_arbol(node.right, level + 1)
            print(' ' * 4 * level + '->', node.value)
            self.imprimir_arbol(node.left, level + 1)

#Inicializar Arbol
abb = ABB()
# Ejemplo:
lista = [1, 2, 3, 4, 5, 6, 7]
arbol = abb.list_to_binary_tree(lista)
```

```python
print("Árbol binario construido a partir de la lista:")
abb.imprimir_arbol(arbol)
```
